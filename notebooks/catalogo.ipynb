## 2. Aplicação

Percorre o dicionário e, para cada tabela, grava no Unity Catalog o comentário da tabela, o
comentário de cada coluna e, onde houver, a chave primária e as estrangeiras.

```python
def aplicar(nome, t):
    tabela = f"workspace.{nome}"
    esc = lambda s: s.replace("'", "\\'")   # apóstrofo dentro do texto quebraria o SQL

    # Comentário da tabela, com o grão na frente
    spark.sql(f"COMMENT ON TABLE {tabela} IS '{esc('Grão: ' + t['grao'] + '. ' + t['desc'])}'")

    # Comentário de cada coluna
    for coluna, texto in t["colunas"].items():
        spark.sql(f"ALTER TABLE {tabela} ALTER COLUMN {coluna} COMMENT '{esc(texto)}'")

    # Chave primária: NOT NULL nas colunas, remove a anterior (e as FKs que dependem dela) e cria
    if "pk" in t:
        for c in t["pk"]:
            spark.sql(f"ALTER TABLE {tabela} ALTER COLUMN {c} SET NOT NULL")
        pk = "pk_" + nome.split(".")[1]
        spark.sql(f"ALTER TABLE {tabela} DROP CONSTRAINT IF EXISTS {pk} CASCADE")
        spark.sql(f"ALTER TABLE {tabela} ADD CONSTRAINT {pk} PRIMARY KEY ({', '.join(t['pk'])})")

    # Chaves estrangeiras: a coluna tem o mesmo nome nas duas tabelas
    for coluna, ref in t.get("fk", {}).items():
        fk = f"fk_{nome.split('.')[1]}_{coluna}"
        spark.sql(f"ALTER TABLE {tabela} DROP CONSTRAINT IF EXISTS {fk}")
        spark.sql(f"ALTER TABLE {tabela} ADD CONSTRAINT {fk} FOREIGN KEY ({coluna}) REFERENCES workspace.{ref}({coluna})")

# Dimensões antes dos fatos: a FK exige que a tabela referenciada exista
for nome, t in catalogo.items():
    aplicar(nome, t)
    print(f"{nome}: {len(t['colunas'])} colunas" + (f", PK {t['pk']}" if "pk" in t else "") + (f", {len(t['fk'])} FKs" if "fk" in t else ""))
```
